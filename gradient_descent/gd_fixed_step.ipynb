### 梯度下降使用固定学习率 $\alpha$

若梯度满足Lipschitz条件，$||\nabla f(x)-\nabla f(y)||_2\le L||x-y||_2$, 则称f为$L$-光滑函数;

在此基础，有下降引理： $f(y)\le f(x)+\nabla f(x)^T(y-x)+\frac{L}{2}||y-x||_2^2$

函数f 凸性： 对任意 $x,y$ 有 $f(y)\ge f(x)+\nabla f(x)^T(y-x)$


在$x_k$处，利用下降引理构造局部上界：
$Q_k(x)=f(x_k)+\nabla f(x_k)^T(x-x_k)+\frac{1}{2\alpha_k} ||x-x_k||_2^2$

求梯度后，得到： $x_{k+1}=x_k-\alpha_k \nabla f(x_k)$

下降更新公式代入下降引理，$f(x_{k+1})\le f(x_k)-\alpha_k(1-\frac{L \alpha _k}{2})||\nabla f(x_k)||_2^2$

$0<\alpha_k<\frac{2}{L}$能保证下降，但是一般保守取 $\alpha_k=\frac{1}{L}$

下降收敛速度由下降引理和凸性联合推导，与条件数$k=\frac{L}{\mu}$ 相关；条件数越大，收敛越慢。（$\mu$ 是描述强凸的引入参数）


```python
import numpy as np
```

定义函数 $f(x)=\frac{1}{2} x^T A x-b^T x$

```python
A=np.array([[4.0,1.0],[1.0,3.0]])
b=np.array([1.0,2.0])

def f(x):
    return 0.5*x @ A @ x-b @ x
```

对于二次函数，根据光滑性，L取A的特征值，取最大。

```python
eigenvalues=np.linalg.eigvalsh(A)
assert np.all(eigenvalues>0) #断言 A的所有特征值为正

L=eigenvalues[-1]  #取最大特征值
alpha=1.0/L

x=np.array([2.0,-1.0]) #起点
tolerance=1e-10  #容差
max_iter=1000 #最大迭代数
iter=0

for i in range(max_iter):
    grad=A@x-b
    if np.linalg.norm(grad)<tolerance:
        print("梯度 趋近0,无法下降")
        break
    x_next=x-alpha*grad
    #验证 下降保证 f(x_{k+1})<=f(x_k)-||∇f(x_k)||² / (2L)
    upper_bound=f(x)-np.dot(grad,grad)/(2.0*L)
    assert f(x_next)<=upper_bound+1e-12
    x=x_next
    iter+=1

#与最优解比较 直接求解 Ax*=b
x_star=np.linalg.solve(A,b)
assert np.allclose(x,x_star,tolerance)

print(f'最优解 x_star={x_star}')
print(f'最优目标值 f(x_star)={f(x_star)}')
print(f'迭代次数 iter={iter}')
print(f'实际解 x={x}')
print(f'实际目标值 f(x)={f(x)}')

```
